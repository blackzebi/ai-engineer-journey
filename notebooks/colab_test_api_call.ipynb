# Colab → Anthropic API smoke test

Verifies the notebook can call the Anthropic API **without ever printing or hardcoding the key**.

Setup (one-time):
1. Open the **key icon** in the left sidebar (Secrets).
2. Click **Add new secret**, name it `ANTHROPIC_API_KEY`, paste your key as the value.
3. Toggle **Notebook access** on for this notebook.
4. Run the cells below.

```python
from google.colab import userdata

api_key = userdata.get('ANTHROPIC_API_KEY')
assert api_key, "Secret ANTHROPIC_API_KEY not found — add it via the key icon in the sidebar."
print(f"Key loaded, length={len(api_key)} (value not shown)")
```

```python
from anthropic import Anthropic

client = Anthropic(api_key=api_key)

response = client.messages.create(
    model="claude-sonnet-5",
    max_tokens=64,
    messages=[{"role": "user", "content": "Reply with the single word: pong"}],
)

print(response.content[0].text)
```
